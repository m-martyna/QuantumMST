# __Solving the MST secular matrix using the NumPy library__ – classical method

```python
from QuantumMST import scatterer, protein_structure, Solvers, FDM
```

The **`Solvers`** function has been created, allowing you to select the **_`mode`_** (the method used to solve the secular matrix problem). 

In addition to the system created via the **`protein_structure`** class, you must provide the energy resolution **_`nE`_** and specify whether you want detailed results displayed (**_`display_stats`_** and **_`plot`_**).

Below is an example of usage with all details displayed:

```python
system21 = protein_structure(reference_system = 5)
results = Solvers(system21, mode='numpy', nE=100, display_stats = True, plot = True)
```

The function itself simply returns an array with the allowed energies for the given system, which we can use to plot the wavefunctions.

```python
results = Solvers(system21, mode='numpy', nE=100)
print(results)
```

For comparison, a simpler and less accurate finite difference method (**`FDM`**) was implemented to verify the validity of the results obtained from the MST approach.

```python
FDM(system21, resolution = 6000)
```

You can switch the **_`mode`_** to VQD algorithm, but you need to specify a chosen **`backend `**, the number of eigenvalues **`k`** and their pentalties **`d`** for cost function.

```python
from qiskit_aer import AerSimulator

system1 = protein_structure(reference_system = 1)
results = Solvers(system1, mode='vqd', nE=20, backend = AerSimulator(), k = 2, b = [1.5, 1.5], display_stats = True, plot = True)
```
